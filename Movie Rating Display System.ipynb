```python
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating
    def display_movie(self):
        print("Movie:", self.name)
        print("Rating:", self.rating, "/ 5")
#Input
name = input("Enter Movie Name: ")
rating = input("Enter Rating: ")
m = Movie(name, rating)
m.display_movie()
```
