```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def display_price_tag(self):
        print("Product:", self.name)
        print("Price: ₹" + str(self.price))
#Input
name = input("Enter Product Name: ")
price = input("Enter Price: ")
p = Product(name, price)
p.display_price_tag()
```
