```python
class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    def display_contact(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.phone)
#user Input
name = input("Enter Contact Name: ")
phone = input("Enter Phone Number: ")
c = Contact(name, phone)
c.display_contact()
```
